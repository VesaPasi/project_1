```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df1 = pd.read_csv("dataset.csv")
df2 = pd.read_csv("train.csv")
df3 = pd.read_csv("test.csv")
df4 = pd.read_csv("val.csv")
```

```python
row_counts = {
    "dataset.csv": len(df1),
    "train.csv": len(df2),
    "test.csv": len(df3),
    "val.csv": len(df4)
}
```

```python
counts_df = pd.DataFrame(list(row_counts.items()), columns=["File", "RowCount"])
```

```python
plt.figure(figsize=(8, 6))
plt.bar(counts_df["File"], counts_df["RowCount"], color="skyblue")
plt.xlabel("CSV Files")
plt.ylabel("Number of Rows")
plt.title("Row Counts in Each CSV File")
plt.xticks(rotation=45)
plt.show()
```
